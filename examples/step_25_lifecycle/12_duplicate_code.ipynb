# Reject duplicate codes on one state type

Declare two different labels under the same code.

## Prepare the model

```python
from aoa.action_machine.domain import BaseStateEntity, value
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:

    class OrderStateEntity(BaseStateEntity):
        """Intentionally declare an ambiguous code."""

        A = value(code="paid", display_name="Paid")
        B = value(code="paid", display_name="Settled")
except ValueError as error:
    assert "duplicate status code" in str(error)
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid declaration was accepted")
```

## What the result means

Different labels do not distinguish two declarations with the same code. The catalogue requires unique codes and rejects the class while collecting its members. Each code must identify one status unambiguously.
