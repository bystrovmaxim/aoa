# Reject a string in the template

Use the removed string form of state().

## Prepare the model

Run this cell first. It imports the required types and defines the model used by this example. Start with a fresh repository Python kernel so declarations from other examples do not enter this model.

```python
from aoa.action_machine.domain import BaseStateEntity, Lifecycle


class OrderStateEntity(BaseStateEntity):
    """Describe one possible order status."""
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    Lifecycle().state("draft")
except TypeError as error:
    assert 'State must be a state entity' in str(error)
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid declaration was accepted")
```

## What the result means

The template builder requires a status object in `.state()`. A string supplies no declared status identity or description and is refused. Code strings remain valid for constructing a lifecycle instance and requesting transitions.
