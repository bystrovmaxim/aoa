## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
template = Lifecycle()

for attribute in ("current_state", "current_status"):
    try:
        getattr(template, attribute)
    except RuntimeError as error:
        print(type(error).__name__ + ": " + str(error))
    else:
        raise AssertionError("A template unexpectedly has a current state")
```

## What the result means

An unselected template has no current status to return. Access raises the shown error instead of manufacturing a default member. Construct the typed lifecycle with a valid code before reading `current_status`.
