# Distinguish status properties from dictionary properties

The dictionary annotation restriction does not apply to lifecycle statuses.

## Prepare the model

Run this cell first. It imports the required types and defines the model used by this example. Start with a fresh repository Python kernel so declarations from other examples do not enter this model.

```python
from aoa.action_machine.domain import BaseStateEntity, value


class OrderStateEntity(BaseStateEntity):
    """Describe a status using the existing state property contract."""
    labels: list[str]
    DRAFT = value(code="draft", display_name="Draft", labels=["new"])
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
print("Status accepted:", OrderStateEntity.DRAFT.labels)
print("Property type:", type(OrderStateEntity.DRAFT.labels).__name__)
assert OrderStateEntity.by_code("draft") is OrderStateEntity.DRAFT
```

## What the result means

The status shares the explicit value() declaration mechanism, but keeps its existing field-type freedom. Frozen prevents replacing an attribute; it does not recursively freeze a list. Prefer immutable status descriptions when sharing them, and remember that graph export still requires a JSON representation.
