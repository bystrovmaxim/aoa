# Classifier names a value, not a field

The marker sits on the extension's reverse field, so the field is already known:
passing a field name is an unknown keyword and the marker refuses it.

## Prepare the model

Run this cell first. It imports the required types and declares the dictionary the marker
names. Start with a fresh repository Python kernel so declarations from other examples do
not enter this model.

```python
from aoa.action_machine.domain import BaseDomain, BaseEnumEntity, Classifier, value
from aoa.action_machine.intents.entity import entity


class MusicDomain(BaseDomain):
    """Group the music catalogue declarations."""

    name = "music"
    description = "A music catalogue"


@entity(description="Kind of VinylRecordCoreEntity", domain=MusicDomain)
class VinylRecordKindEnumEntity(BaseEnumEntity):
    """Declare the codes the axis borrows."""

    FIRST = value(code="first", display_name="First")
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    Classifier(field="record", variant=VinylRecordKindEnumEntity.FIRST)
except TypeError as error:
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid marker was accepted")

marker = Classifier(variant=VinylRecordKindEnumEntity.FIRST)
print("A value, and no field name:", marker.variant.code, repr(marker))
```

The marker carries the dictionary value and nothing else. The attribute it sits on is the
pair, so a field name could only repeat it — and a typo in that name used to be a build
error. The keyword is gone, and the class body says so at once.
