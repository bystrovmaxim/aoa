# Name exactly one domain

``domain=`` takes one class that is a ``BaseDomain`` subclass; a declaration without one is refused when the class is defined, naming the entity. ``resource=`` is gone, and the decorator refuses it.

Use the repository Python environment and run the cells in order. This notebook has its own setup; no previous notebook is required.

## 1. Import the implementation

```python
from pydantic import Field

from aoa.action_machine.domain import BaseCoreEntity, EntityDecoratorError
from aoa.action_machine.domain.base_domain import BaseDomain
from aoa.action_machine.intents.entity import entity
```

## 2. Declare the model

```python
class StoreDomain(BaseDomain):
    """Group the shop model declarations."""

    name = "store"
    description = "Store"


def declare(domain: object) -> None:
    """Declare one order in ``domain`` and print the refusal."""
    try:

        @entity(description="Customer order", domain=domain)  # type: ignore[arg-type]
        class SalesOrderCoreEntity(BaseCoreEntity):
            """A customer's order."""

            id: str = Field(description="Order identifier")

    except EntityDecoratorError as error:
        print(type(error).__name__ + ":", error)
    else:
        raise AssertionError(f"domain={domain!r} was accepted, but must be refused")
```

## 3. Run and check the observation

```python
declare(None)

try:
    entity(description="Customer order", resource=StoreDomain)  # type: ignore[call-arg]
except TypeError as error:
    print(type(error).__name__ + ":", error)
else:
    raise AssertionError("resource= was accepted")
```

## 4. Explain the result

`domain=` takes one `BaseDomain` subclass. A declaration without one, and the removed `resource=` argument, are refused when the class is defined.
