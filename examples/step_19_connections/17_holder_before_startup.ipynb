## 2. Define the objects used in this case

These definitions give the case an observable result without introducing unrelated business behavior.

```python
class PoolHolder:
    """Expose a pool after service startup."""
    pool = None
    def require(self):
        """Reject use before startup or after shutdown."""
        if self.pool is None:
            raise RuntimeError("Pool is not started")
        return self.pool
```

## 4. Run and check the observation

The assertions test this case’s stated behavior. Printed output is recorded below; expected failures are caught explicitly.

```python
try:
    PoolHolder().require()
except RuntimeError as error:
    assert str(error) == "Pool is not started"
    print(str(error))
else:
    raise AssertionError("An empty holder returned a pool")
```

## 5. Explain the result

A holder is an application object with a checked accessor. The factory reads it after startup; an early read produces an explicit error instead of an accidental attribute error.
