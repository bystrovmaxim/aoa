# Refuse Detail

The property annotation is checked when the dictionary class is created.

## Prepare the model

Run this cell first. It imports the required types and defines the model used by this example. Start with a fresh repository Python kernel so declarations from other examples do not enter this model.

```python
from pydantic import BaseModel

from aoa.action_machine.domain import BaseEnumEntity, value


class Detail(BaseModel):
    """Describe a nested model that is outside the property contract."""
    count: int = 1
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    class InvalidEnumEntity(BaseEnumEntity):
        """Attempt to declare an unsupported property type."""
        detail: Detail
        ITEM = value(code="item", display_name="Item")
except ValueError as error:
    print(str(error))
    assert "field 'detail' is Detail" in str(error)
else:
    raise AssertionError("The unsupported property was accepted")
```

## What the result means

Nested Pydantic models are outside the supported dictionary-property types. The error names the class, field and unsupported type. This check runs before the value arguments are validated, so it does not depend on supplying an example of the forbidden value.
