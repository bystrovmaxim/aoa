# A storage that is not PostgreSQL takes no pool

``pool=`` and ``connection_params=`` belong to a PostgreSQL storage. A storage of another kind knows no pool, and giving it one is refused when it is built.

Use the repository Python environment and run the cells in order. This notebook has its own setup; no previous notebook is required.

## 1. Import the implementation

```python
!pip install -q "aoa-action-machine[postgres]"
```

```python
from aoa.action_machine.intents.entities import entities
from aoa.action_machine.intents.meta import meta
from aoa.action_machine.resources import BaseStorage
from aoa.action_machine.resources.postgres import PostgresPool
```

## 2. Declare the model

```python
@meta(description="An in-memory cache of the shop")
@entities()
class ShopCacheStorage(BaseStorage):
    """Keeps its data in memory; there is no pool to take."""

    @classmethod
    def get_wrapper_class(cls) -> None:
        """No wrapper: nothing to protect."""
        return None
```

## 3. Run and check the observation

```python
for arguments in ({"pool": PostgresPool(pool=object())}, {"connection_params": "postgresql://example.invalid/shop"}):
    try:
        ShopCacheStorage(**arguments)
    except TypeError as error:
        print(f"{sorted(arguments)[0]}= refused with {type(error).__name__}")
    else:
        raise AssertionError("A pool was given to a storage that knows none")
```

## 4. Explain the result

`pool=` and `connection_params=` belong to a PostgreSQL storage; a storage of another kind takes neither, and Python refuses the unknown arguments when it is built.
