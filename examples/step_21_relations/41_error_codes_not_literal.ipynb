# Classifier accepts a value, not a list

Pass a list of code strings to Classifier.

## Prepare the model

Run this cell first. It imports the required types and defines the model used by this example. Start with a fresh repository Python kernel so declarations from other examples do not enter this model.

```python
from aoa.action_machine.domain import BaseDomain, BaseEnumEntity, Classifier, value
from aoa.action_machine.intents.entity import entity


class MediaDomain(BaseDomain):
    """Group the declarations this experiment needs."""

    name = "media"
    description = "Media"


@entity(description="Media kind", domain=MediaDomain)
class MediaKindEnumEntity(BaseEnumEntity):
    """Declare the codes an axis borrows, in the class body."""

    VINYL = value(code="vinyl", display_name="Vinyl")
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    Classifier("media", ["first"])  # a list is not a value
except TypeError as error:
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid marker was accepted")

marker = Classifier(variant=MediaKindEnumEntity.VINYL)
print("A dictionary value, not a code list:", marker.variant.code)
print("Declared:", MediaKindEnumEntity.codes())
```

## What the result means

`Classifier` names one declared dictionary object, not a list of code strings. The invalid list fails immediately; the valid marker keeps `MediaKindEnumEntity.VINYL`. The head’s dictionary type supplies the vocabulary.
