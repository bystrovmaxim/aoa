# Keep a status object immutable

Try to rename a status after creating it.

## Prepare the model

Run this cell first. It imports the required types and defines the model used by this example. Start with a fresh repository Python kernel so declarations from other examples do not enter this model.

```python
from pydantic import ValidationError

from aoa.action_machine.domain import BaseStateEntity


class OrderStateEntity(BaseStateEntity):
    """Describe one possible order status."""


status = OrderStateEntity(code="paid", display_name="Paid")
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    status.display_name = "Renamed"
except ValidationError as error:
    detail = error.errors()[0]
    print(type(error).__name__ + ": " + str(detail["loc"][0]) + " — " + detail["msg"])
else:
    raise AssertionError("The invalid declaration was accepted")
```

## What the result means

The frozen model prevents replacing the status’s label, so the original text remains. It does not recursively freeze arbitrary nested objects. The separate status-property example demonstrates that distinction with a list.
