# Reject a string in the template

Use the removed string form of state().


## Declare the model

Run these definitions first. This notebook is independent of the other examples.


```python
from aoa.action_machine.domain import BaseStateEntity, Lifecycle, StateVocabulary


class OrderStateEntity(BaseStateEntity):
    """Describe one possible order status."""
```

## Run the case


```python
try:
    Lifecycle().state("draft")
except TypeError as error:
    assert 'State must be a state entity' in str(error)
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid declaration was accepted")
```

## What the result means

Template declarations now require status objects. Write state(DRAFT), with the label in DRAFT.display_name. The old two-string form state("draft", "Draft") is also unsupported.
