# A template has no current status

Ask a template for the current code and the current status object.


```python
from aoa.action_machine.domain import Lifecycle

template = Lifecycle()
```

## Read the runtime properties


```python
for attribute in ("current_state", "current_status"):
    try:
        getattr(template, attribute)
    except RuntimeError as error:
        print(type(error).__name__ + ": " + str(error))
    else:
        raise AssertionError("A template unexpectedly has a current state")
```

## What the result means

A template describes possible states and transitions, but selects no current state. Both runtime properties therefore raise RuntimeError. Construct a specialized lifecycle with a valid code before accessing them; inspecting the template through get_states() remains available.
