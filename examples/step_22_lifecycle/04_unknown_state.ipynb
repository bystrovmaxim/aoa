# Reject an unknown current state

Construct a lifecycle using a code absent from its template.


## Declare the model

Run these definitions first. This notebook is independent of the other examples.


```python
from aoa.action_machine.domain import BaseStateEntity, Lifecycle


class OrderStateEntity(BaseStateEntity):
    """Describe one possible order status."""

    DRAFT = OrderStateEntity(code="draft", display_name="Draft")  # noqa: F821
    PAID = OrderStateEntity(code="paid", display_name="Payment received")  # noqa: F821



class OrderLifecycle(Lifecycle[OrderStateEntity]):
    """Allow a draft order to become paid."""


    _template = (
        Lifecycle()
        .state(OrderStateEntity.DRAFT).to(OrderStateEntity.PAID).initial()
        .state(OrderStateEntity.PAID).final()
    )

from aoa.action_machine.domain.lifecycle import InvalidStateError
```

## Run the case


```python
try:
    OrderLifecycle("missing")
except InvalidStateError as error:
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("An unknown state was accepted")
```

## What the result means

The constructor checks membership in the template. A known final code such as paid is valid at construction; construction does not replay a path from the initial state.
