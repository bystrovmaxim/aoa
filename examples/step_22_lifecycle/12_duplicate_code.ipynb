# Reject duplicate codes in one vocabulary

Declare two different labels under the same code.


## Declare the model

Run these definitions first. This notebook is independent of the other examples.


```python
from aoa.action_machine.domain import BaseStateEntity
```

## Run the case


```python
try:
    class OrderStateEntity(BaseStateEntity):
        """Intentionally declare an ambiguous code."""

        A = OrderStateEntity(code="paid", display_name="Paid")  # noqa: F821
        B = OrderStateEntity(code="paid", display_name="Settled")  # noqa: F821
except ValueError as error:
    assert "duplicate status code" in str(error)
    print(type(error).__name__ + ": " + str(error))
else:
    raise AssertionError("The invalid declaration was accepted")
```

## What the result means

A code identifies one status within a vocabulary. The class cannot be created when two attributes use that code, even if their labels differ. Different lifecycle vocabularies may reuse a code.
