The class name is no longer replaced by a hidden declaration function.

```python
from aoa.action_machine.domain import BaseEnumEntity
```

## Run the example

The following cell performs the operation described above. Its assertions check the result; any expected refusal is caught explicitly.

```python
try:
    class OldEnumEntity(BaseEnumEntity):
        """Attempt to use the class before it exists."""
        EUR = OldEnumEntity(code="eur", display_name="Euro")
except NameError:
    print("Old declaration: NameError")
else:
    raise AssertionError("The removed declaration alias was accepted")
```

## What the result means

In this scope the name OldEnumEntity has not been bound when its body executes. Import value and write EUR = value(...) instead. A same-named binding in an outer scope would follow ordinary Python lookup; the framework no longer supplies a special class-name alias.
